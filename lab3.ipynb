№ 1. Возьмите реализацию класса UnorderedList из лекционных материалов и выполните
следующие доработки:
1. Реализуйте оставшиеся методы этого класса append(item), index(item),
insert(pos,item), pop(), pop(pos). Требования к работе этих методов посмотрите в
лекционной презентации.
2. Реализуйте метод __str__, чтобы содержимое списка отображалось в Python-стиле (в
квадратных скобках).
3. Реализуйте метод “среза”. Он должен принимать два параметра (start и stop) и
возвращать копию списка, начиная с позиции start и заканчивая (но не включая!)
позицией stop.
Все выполненные доработки должны быть протестированы.

````python
class Node:
    def __init__(self, initdata):
        self.data = initdata
        self.next = None

    def getData(self):
        return self.data

    def getNext(self):
        return self.next

    def setData(self, newdata):
        self.data = newdata

    def setNext(self, newnext):
        self.next = newnext


class UnorderedList:

    def __init__(self):
        self.head = None

    def isEmpty(self):
        return self.head is None

    def add(self, item):
        temp = Node(item)
        temp.setNext(self.head)
        self.head = temp

    def size(self):
        current = self.head
        count = 0
        while current is not None:
            count += 1
            current = current.getNext()
        return count

    def search(self, item):
        current = self.head
        found = False
        while current is not None and not found:
            if current.getData() == item:
                found = True
            else:
                current = current.getNext()
        return found

    def remove(self, item):
        current = self.head
        previous = None
        found = False
        while not found:
            if current.getData() == item:
                found = True
            else:
                previous = current
                current = current.getNext()
            if current is None:
                raise ValueError("Item not found in list")
        if previous is None:
            self.head = current.getNext()
        else:
            previous.setNext(current.getNext())

    # Доработки ниже

    def append(self, item):
        """Добавляет элемент в конец списка"""
        temp = Node(item)
        if self.head is None:
            self.head = temp
        else:
            current = self.head
            while current.getNext() is not None:
                current = current.getNext()
            current.setNext(temp)

    def index(self, item):
        """Возвращает индекс первого вхождения item, иначе ValueError"""
        current = self.head
        idx = 0
        while current is not None:
            if current.getData() == item:
                return idx
            current = current.getNext()
            idx += 1
        raise ValueError("Item not found in list")

    def insert(self, pos, item):
        """Вставляет элемент item на позицию pos"""
        if pos < 0 or pos > self.size():
            raise IndexError("Index out of range")
        temp = Node(item)
        if pos == 0:
            temp.setNext(self.head)
            self.head = temp
        else:
            previous = None
            current = self.head
            idx = 0
            while idx < pos:
                previous = current
                current = current.getNext()
                idx += 1
            previous.setNext(temp)
            temp.setNext(current)

    def pop(self, pos=None):
        """Удаляет и возвращает элемент по индексу pos, либо последний, если pos не указан"""
        sz = self.size()
        if sz == 0:
            raise IndexError("pop from empty list")
        if pos is None:
            pos = sz - 1
        if pos < 0 or pos >= sz:
            raise IndexError("pop index out of range")
        previous = None
        current = self.head
        idx = 0
        while idx < pos:
            previous = current
            current = current.getNext()
            idx += 1
        if previous is None:
            self.head = current.getNext()
        else:
            previous.setNext(current.getNext())
        return current.getData()

    def __str__(self):
        """Строковое представление списка как [a, b, c]"""
        result = []
        current = self.head
        while current is not None:
            result.append(repr(current.getData()))
            current = current.getNext()
        return '[' + ', '.join(result) + ']'

    def slice(self, start, stop):
        """Возвращает копию списка, начиная с позиции start и заканчивая (не включая) stop"""
        if start < 0 or stop < 0 or start > self.size() or stop > self.size() or start > stop:
            raise IndexError("Invalid slice indices")
        new_list = UnorderedList()
        current = self.head
        idx = 0
        while current is not None and idx < stop:
            if idx >= start:
                new_list.append(current.getData())
            current = current.getNext()
            idx += 1
        return new_list


# Тесты

mylist = UnorderedList()

print("Список после добавления элементов (add):", mylist)
mylist.add(31)
mylist.add(77)
mylist.add(17)
mylist.add(93)
mylist.add(26)
mylist.add(54)
print("Список после добавления элементов (add):", mylist)
print("Размер списка:", mylist.size())
print("Результат поиска 93:", mylist.search(93))
print("Результат поиска 100:", mylist.search(100))

mylist.append(100)
print("Список после append(100):", mylist)
print("Индекс элемента 93:", mylist.index(93))
try:
    print("Индекс элемента 999:", mylist.index(999))
except ValueError as e:
    print("Ошибка поиска индекса 999:", e)

mylist.insert(2, 42)
print("Список после вставки 42 на позицию 2:", mylist)

popped_item = mylist.pop()
print("Извлечённый элемент (без указания позиции):", popped_item)
print("Список после извлечения элемента:", mylist)

popped_item_pos = mylist.pop(2)
print("Извлечённый элемент (позиция 2):", popped_item_pos)
print("Список после извлечения элемента на позиции 2:", mylist)

sliced_list = mylist.slice(1, 4)
print("Срез списка с позиции 1 по 4:", sliced_list)

empty_slice = mylist.slice(0, 0)
print("Срез с 0 по 0 (должен быть пустым):", empty_slice)

mylist.remove(77)
print("Список после удаления 77:", mylist)
````

№ 2. Напишите функцию move_to_front для чтения строк с клавиатуры и сохранения их в
связном списке без повторов. Пользователь последовательно вводит с клавиатуры n строк.
При вводе строки, которая раньше не встречалась, вставьте ее в начало списка. При вводе
уже внесенной строки удалите ее из списка и снова вставьте в начало.
Такая функция реализует стратегию «сдвиг в начало», которая применяется в кешировании,
при сжатии данных и т.д.

````python
def move_to_front(n):
    ul = UnorderedList()
    for _ in range(n):
        s = input(f"Введите строку {_+1}: ")
        if ul.search(s):
            ul.remove(s)
        ul.add(s)
        print("Текущее состояние списка:", ul)
    return ul

print(move_to_front(5))
````

№ 3. Напишите функцию, которая принимает в качестве аргумента первый узел односвязного
списка, обращает порядок элементов списка (с разрушением исходного) и возвращает в
качестве результата первый узел полученного списка. Разработайте два варианта решения
этой задачи: 1) итеративный и 2) рекурсивный.

````python
def reverse_iterative(head):
    prev = None
    current = head
    while current is not None:
        next_node = current.getNext()
        current.setNext(prev)
        prev = current
        current = next_node
    return prev

def reverse_recursive(head, prev=None):
    if head is None:
        return prev
    next_node = head.getNext()
    head.setNext(prev)
    return reverse_recursive(next_node, head)

def print_list(head):
    current = head
    result = []
    while current:
        result.append(current.data)
        current = current.next
    print(" -> ".join(map(str, result)))

head = Node(1)
head.next = Node(2)
head.next.next = Node(3)
head.next.next.next = Node(4)
head.next.next.next.next = Node(5)

print("Исходный список:")
print_list(head)

# Итеративное обращение
reversed_head_iterative = reverse_iterative(head)
print("\nСписок после итеративного обращения:")
print_list(reversed_head_iterative)

# Рекурсивное обращение
reversed_head_recursive = reverse_recursive(reversed_head_iterative)
print("\nСписок после рекурсивного обращения:")
print_list(reversed_head_recursive)
````

№ 4. Используя реализацию связного списка из задания № 1 (класс UnorderedList), реализуйте
следующие линейные структуры (необходимый набор методов посмотрите в лекционной

1. Стек.

Проведите тестирование реализованных линейных структур.
Продумайте и проведите показательный эксперимент, сравнивающий производительность
стека, очереди или дека, основанных на списках Python, с реализациями на основе
связанного списка. Например, для сравнения можно взять какую-либо операцию
выбранной линейной структуры.

````python
class Stack:
    def __init__(self):
        self.items = UnorderedList()

    def isEmpty(self):
        return self.items.isEmpty()

    def push(self, item):
        self.items.add(item)

    def pop(self):
        if self.items.isEmpty():
            raise IndexError("pop from empty stack")
        return self.items.pop(0)  # снимаем с головы

    def peek(self):
        if self.items.isEmpty():
            raise IndexError("peek from empty stack")
        return self.items.head.getData()

    def size(self):
        return self.items.size()

    def __str__(self):
        return self.items.__str__()


class Queue:
    def __init__(self):
        self.items = UnorderedList()

    def enqueue(self, item):
        # Вставка в хвост
        self.items.append(item)

    def dequeue(self):
        if self.items.isEmpty():
            raise IndexError("dequeue from empty queue")
        return self.items.pop(0)

    def isEmpty(self):
        return self.items.isEmpty()

    def size(self):
        return self.items.size()

    def __str__(self):
        return self.items.__str__()


class Deque:
    def __init__(self):
        self.items = UnorderedList()

    def addFront(self, item):
        self.items.add(item)

    def addRear(self, item):
        self.items.append(item)

    def removeFront(self):
        if self.items.isEmpty():
            raise IndexError("removeFront from empty deque")
        return self.items.pop(0)  # снимаем с головы

    def removeRear(self):
        if self.items.isEmpty():
            raise IndexError("removeRear from empty deque")
        return self.items.pop()  # снимаем с хвоста

    def isEmpty(self):
        return self.items.isEmpty()

    def size(self):
        return self.items.size()

    def __str__(self):
        return self.items.__str__()


# Тесты:

print("Тест стека:")
stack = Stack()
stack.push(1)
stack.push(2)
stack.push(3)
print(f"Исходный стек: {stack}")
print(f"Пустой: {stack.isEmpty()}")
print(f"Извлечен: {stack.pop()}")
print(f"Стек после извлечения: {stack}")
print(f"Вершина стека: {stack.peek()}")
print(f"Размер стека: {stack.size()}")

print("\nТест очереди:")
queue = Queue()
queue.enqueue(1)
queue.enqueue(2)
queue.enqueue(3)
print(f"Исходная очередь: {queue}")
print(f"Пустая: {queue.isEmpty()}")
print(f"Извлечен: {queue.dequeue()}")
print(f"Очередь после извлечения: {queue}")
print(f"Размер очереди: {queue.size()}")

print("\nТест дека:")
deque = Deque()
deque.addFront(1)
deque.addRear(2)
deque.addFront(0)
deque.addRear(3)
print(f"Исходный дек: {deque}")
removed_front = deque.removeFront()
print(f"Удален с фронта: {removed_front}")
print(f"Дек после удаления с фронта: {deque}")
removed_rear = deque.removeRear()
print(f"Удален с хвоста: {removed_rear}")
print(f"Дек после удаления с хвоста: {deque}")
print(f"Пустой: {deque.isEmpty()}")
print(f"Размер дека: {deque.size()}")
````

№ 5. Реализуйте класс DoubleList для построения двух связных списков, в которых каждый узел
содержит ссылку на узел, предшествующий ему в списке, и ссылку на следующий узел.
Класс должен иметь следующие методы: проверка списка на пустоту, поиск элемента,
размер списка, вставка в начало, вставка в конец, удаление из начала, удаление из конца,
вставка перед указанным узлом, вставка после указанного узла, удаление указанного узла.
Проведите тестирование методов реализованного класса.

````python
class DoubleNode:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None

class DoubleList:
    def __init__(self):
        self.head = None
        self.tail = None

    def isEmpty(self):
        return self.head is None

    def size(self):
        count = 0
        current = self.head
        while current:
            count += 1
            current = current.next
        return count

    def search(self, item):
        current = self.head
        while current:
            if current.data == item:
                return current
            current = current.next
        return None

    def insertFront(self, item):
        new_node = DoubleNode(item)
        new_node.next = self.head
        if self.head:
            self.head.prev = new_node
        self.head = new_node
        if self.tail is None:
            self.tail = new_node

    def insertRear(self, item):
        new_node = DoubleNode(item)
        new_node.prev = self.tail
        if self.tail:
            self.tail.next = new_node
        self.tail = new_node
        if self.head is None:
            self.head = new_node

    def removeFront(self):
        if self.isEmpty():
            raise IndexError("removeFront from empty list")
        data = self.head.data
        self.head = self.head.next
        if self.head:
            self.head.prev = None
        else:
            self.tail = None
        return data

    def removeRear(self):
        if self.isEmpty():
            raise IndexError("removeRear from empty list")
        data = self.tail.data
        self.tail = self.tail.prev
        if self.tail:
            self.tail.next = None
        else:
            self.head = None
        return data

    def insertBefore(self, node, item):
        if node is None:
            raise ValueError("insertBefore: given node is None")
        new_node = DoubleNode(item)
        new_node.prev = node.prev
        new_node.next = node
        if node.prev:
            node.prev.next = new_node
        else:
            self.head = new_node
        node.prev = new_node

    def insertAfter(self, node, item):
        if node is None:
            raise ValueError("insertAfter: given node is None")
        new_node = DoubleNode(item)
        new_node.next = node.next
        new_node.prev = node
        if node.next:
            node.next.prev = new_node
        else:
            self.tail = new_node
        node.next = new_node

    def removeNode(self, node):
        if node is None:
            raise ValueError("removeNode: given node is None")
        if node.prev:
            node.prev.next = node.next
        else:
            self.head = node.next
        if node.next:
            node.next.prev = node.prev
        else:
            self.tail = node.prev
        return node.data

    def __str__(self):
        result = []
        current = self.head
        while current:
            result.append(repr(current.data))
            current = current.next
        return '[' + ', '.join(result) + ']'

# Тесты

dl = DoubleList()
print("Пустой?", dl.isEmpty())
dl.insertFront(10)
dl.insertRear(20)
dl.insertFront(5)
dl.insertRear(30)
print("После добавления элементов:", dl)
print("Размер:", dl.size())

node_20 = dl.search(20)
dl.insertBefore(node_20, 15)
print("После вставки 15 перед 20:", dl)

node_10 = dl.search(10)
dl.insertAfter(node_10, 12)
print("После вставки 12 после 10:", dl)

print("Удаление из начала:", dl.removeFront())
print("После удаления из начала:", dl)
print("Удаление из конца:", dl.removeRear())
print("После удаления из конца:", dl)

node_12 = dl.search(12)
print("Удаление узла с 12:", dl.removeNode(node_12))
print("После удаления узла 12:", dl)

print("Пустой?", dl.isEmpty())
print("Осталось:", dl)
````

№ 6. В HTML у каждого тега существуют открывающая и закрывающая формы, которые должны
быть сбалансированы, чтобы адекватно описывать веб-документ. Вот пример очень
простого веб-документа:

```
<html>
    <head>
        <title>Example</title>
    </head>
    <body>
        <h1>Hello, world</h1>
    </body>
</html>
```

Напишите программу, которая, используя структуру данных стек, проверит HTML
документ на соответствие открывающих и закрывающих тегов.
За основу решения можно взять алгоритм проверки скобочной последовательности из
лекционных материалов.

````python
def check_html_tags(html_string):
    stack = Stack()
    i = 0
    while i < len(html_string):
        if html_string[i] == "<":
            close_pos = html_string.find(">", i + 1)
            if close_pos == -1:
                # Нет закрывающего символа '>'
                return False
            tag_content = html_string[i + 1:close_pos]
            if not tag_content.startswith("/"):
                # Открывающий тег
                stack.push(tag_content)
            else:
                # Закрывающий тег
                if stack.isEmpty():
                    return False
                last_tag = stack.pop()
                if last_tag != tag_content[1:]:
                    return False
            i = close_pos
        i += 1
    return stack.isEmpty()

# Тесты
html1 = "<html><head><title>Example</title></head><body><h1>Hello, world</h1></body></html>"
html2 = "<html><head><title>Example</head><body><h1>Hello</h1></body></html>"  # некорректный

print(check_html_tags(html1))  # True
print(check_html_tags(html2))  # False
````

№ 7. Стековый калькулятор - это устройство (реальное или виртуальное), которое вычисляет
значения арифметических выражений, записанных в постфиксной форме (также она носит
название Обратная польская запись — Википедия (wikipedia.org)). Стековым он называется
так как действует по следующим правилам: если нам встречается число - мы должны
поместить его в стек; если знак операции - мы должны достать из стека столько чисел,
сколько необходимо для выполнения данной операции и выполнить ее.
Именно так и работает т.н. математический сопроцессор (например, в INtel8086+AMD64)-
часть ЦП, отвечающая за операции над числами с плавающей запятой.
Реализуйте стековый калькулятор. Напишите программу, которая читает выражение в
обратной польской нотации и выводит получившееся значение или пишет, что выражение
составлено не корректно (если оно некорректно).

````python
def postfix_calculator(expr):
    stack = Stack()
    ops = {'+', '-', '*', '/'}
    tokens = expr.split()

    for tok in tokens:
        if tok.isdigit() or (tok.startswith('-') and tok[1:].isdigit()):
            stack.push(int(tok))
        elif tok in ops:
            if stack.size() < 2:
                return "Ошибка: некорректное выражение"
            right = stack.pop()
            left = stack.pop()
            if tok == '+':
                stack.push(left + right)
            elif tok == '-':
                stack.push(left - right)
            elif tok == '*':
                stack.push(left * right)
            elif tok == '/':
                if right == 0:
                    return "Ошибка: деление на ноль"
                stack.push(left / right)
        else:
            return "Ошибка: некорректное выражение"

    if stack.size() != 1:
        return "Ошибка: некорректное выражение"
    return stack.pop()


# Примеры использования
examples = [
    "3 4 + 2 * 7 /",      # (3+4)*2/7 = 2.0
    "5 1 2 + 4 * + 3 -",  # 5 + ((1+2)*4) - 3 = 14
    "10 2 /",             # 10/2 = 5.0
    "3 4 + 2 * 7",        # Некорректное выражение
    "1 0 /",              # Деление на ноль
    "2 3 ^",              # Некорректное выражение (неподдерживаемая операция)
    "-3 -4 *",            # -3 * -4 = 12
]

for expr in examples:
    print(f"Выражение: {expr} -> {postfix_calculator(expr)}")
````
