Изучите сортировку слиянием и реализуйте ее в виде функции merge_sort(some_list). Функция принимает список и сортирует его по возрастанию.

```python
def merge_sort(some_list):
    a = some_list[:]  # работаем с копией
    n = len(a)
    width = 1

    def merge(left, right):
        i = j = 0
        result = []
        while i < len(left) and j < len(right):
            if left[i] <= right[j]:
                result.append(left[i])
                i += 1
            else:
                result.append(right[j])
                j += 1
        result.extend(left[i:])
        result.extend(right[j:])
        return result

    # Итеративное слияние блоков размера width
    while width < n:
        result = []
        for i in range(0, n, 2 * width):
            left = a[i:i + width]
            right = a[i + width:i + 2 * width]
            result. extend(merge(left, right))
        a = result
        width *= 2

    return a
```

```python
def merge_sort_with_passes(arr):
    # passes — список состояний массива после каждого прохода (уровня).
    a = arr[:]  # работаем с копией
    n = len(a)
    width = 1
    passes = []

    def merge(left, right):
        i = j = 0
        result = []
        while i < len(left) and j < len(right):
            if left[i] <= right[j]:
                result.append(left[i])
                i += 1
            else:
                result.append(right[j])
                j += 1
        result.extend(left[i:])
        result.extend(right[j:])
        return result

    # Итеративное слияние блоков размера width
    while width < n:
        result = []
        for i in range(0, n, 2 * width):
            left = a[i:i + width]
            right = a[i + width:i + 2 * width]
            result.extend(merge(left, right))
        a = result
        passes.append(a[: ])  # сохраняем состояние после прохода
        width *= 2

    return a, passes
```

```python
arr1 = [7, 3, 9, 4, 2, 5, 6, 1, 8]
sorted_list, passes = merge_sort_with_passes(arr1)

print("Исходный:", arr1)
for k, state in enumerate(passes, 1):
    print(f"После прохода {k}:", state)
print("Итог:", sorted_list)
```

## Временная сложность
### Сортировка слиянием имеет временную сложность O(n log n) во всех случаях — это одно из её главных преимуществ.

1. Деление массива (log n уровней):
Алгоритм делит массив пополам на каждом шаге

Количество уровней деления = ⌈log₂(n)⌉

Например, для n=8: 8 → 4 → 2 → 1 (3 уровня = log₂8)

2. Слияние на каждом уровне (n операций):
На каждом уровне мы обрабатываем все n элементов

Слияние двух отсортированных списков требует линейного времени O(n)

Итого: log₂(n) уровней × n операций на уровень = O(n log n)

## Пространственная сложность
### Итеративная (bottom-up) реализация имеет пространственную сложность O(n).

1. Копия исходного массива: O(n)

a = arr[:]  - создаём копию
2. Временные массивы для слияния: O(n)

left = a[i:i + width]     - срезы массива

right = a[i + width:...]  - суммарно O(n) на уровень

result = []               - результат слияния O(n)

3. Стек вызовов: O(1)
Используем циклы while и for, без рекурсии

Нет рекурсивных вызовов → нет затрат на стек



```python
import random

def random_list(n):
    # a.  Список случайных чисел
    return [random. randint(0, n * 10) for _ in range(n)]


def almost_sorted(n):
    # b. Почти отсортированный список (один элемент в середине списка находится не на своем месте)
    arr = list(range(n))  # [0, 1, 2, ..., n-1]
    if n > 1:
        mid = n // 2
        # Меняем средний элемент с одним из соседних
        arr[mid], arr[mid - 1] = arr[mid - 1], arr[mid]
    return arr


def almost_sorted_reverse(n):
    # c. Почти отсортированный в обратную сторону список (один элемент в середине списка находится не на своем месте)
    arr = list(range(n - 1, -1, -1))  # [n-1, n-2, ..., 1, 0]
    if n > 1:
        mid = n // 2
        # Меняем средний элемент с одним из соседних
        arr[mid], arr[mid - 1] = arr[mid - 1], arr[mid]
    return arr


def half_zeros_half_ones(n):
    # d. Половина данных равна 0, другая половина 1
    half = n // 2
    arr = [0] * half + [1] * (n - half)
    random.shuffle(arr)  # перемешиваем для случайного порядка
    return arr


def sorted_95_random_5(n):
    # e. Первые 95% списка упорядочены, а последние 5% содержат случайные значения
    sorted_part = int(n * 0.95)
    random_part = n - sorted_part

    arr = list(range(sorted_part))  # упорядоченная часть
    arr.extend([random.randint(0, n * 10) for _ in range(random_part)])  # случайная часть
    return arr


def within_10_positions(n):

    # f. Все элементы находятся в пределах 10 позиций от окончательного места в массиве
    arr = list(range(n))  # начинаем с отсортированного массива

    for i in range(n):
        # Смещаем элемент на случайную величину в пределах [-10, 10]
        offset = random.randint(-10, 10)
        new_pos = i + offset

        # Ограничиваем позицию в пределах массива
        new_pos = max(0, min(n - 1, new_pos))

        # Меняем элементы местами
        arr[i], arr[new_pos] = arr[new_pos], arr[i]

    return arr
```

```python
import time
import matplotlib. pyplot as plt

# Размеры входных данных
sizes = [100, 500, 1000, 2000, 4000, 6000, 8000, 10000]

# Список датасетов с описаниями и функциями-генераторами
datasets = [
    ("a.  Список случайных чисел", random_list),
    ("b. Почти отсортированный список", almost_sorted),
    ("c. Почти отсортированный в обратную сторону", almost_sorted_reverse),
    ("d. Половина нулей, половина единиц", half_zeros_half_ones),
    ("e. 95% отсортированы, 5% случайные", sorted_95_random_5),
    ("f. В пределах 10 позиций от места", within_10_positions)
]

# Словарь для хранения всех результатов
all_results = {}

# Проводим эксперименты для каждого типа данных
for name, generator in datasets:
    print(f"Тестирование:  {name}")
    times = []

    for n in sizes:
        # Генерируем данные
        data = generator(n)

        # Замеряем время сортировки
        start = time.time()
        merge_sort(data)
        elapsed = time.time() - start

        times.append(elapsed)
        print(f"  n={n}:  {elapsed:.6f} сек")

    all_results[name] = times

    # Строим график для текущего типа данных
    plt. figure(figsize=(10, 6))
    plt.plot(sizes, times, marker='o', linewidth=2, markersize=8)
    plt.xlabel("Размер входных данных (n)", fontsize=12)
    plt.ylabel("Время выполнения (секунды)", fontsize=12)
    plt.title(name, fontsize=14, fontweight='bold')
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

# Строим сводный график со всеми типами данных
plt.figure(figsize=(12, 8))
for name, times in all_results. items():
    plt.plot(sizes, times, marker='o', linewidth=2, markersize=6, label=name)

plt.xlabel("Размер входных данных (n)", fontsize=12)
plt.ylabel("Время выполнения (секунды)", fontsize=12)
plt.title("Сравнение времени работы merge_sort на разных типах данных", fontsize=14, fontweight='bold')
plt.legend(fontsize=9, loc='upper left')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
final_times = {name: times[-1] for name, times in all_results.items()}
fastest = min(final_times, key=final_times.get)
slowest = max(final_times, key=final_times. get)

print(f"\nДля n=10000:")
print(f"Самый быстрый: {fastest} ({final_times[fastest]:.6f} сек)")
print(f"Самый медленный: {slowest} ({final_times[slowest]:.6f} сек)")
```

### Почему почти отсортированный быстрее?
Кэш-локальность: последовательный доступ к памяти

Предсказание ветвлений: процессор правильно угадывает результаты сравнений

Меньше перестановок: элементы уже близки к нужным местам

### Почему случайные числа медленнее?
Непредсказуемые сравнения: процессор часто ошибается в предсказаниях

Плохая кэш-локальность: больше обращений к медленной RAM

Максимум операций: больше копирований и перестановок
